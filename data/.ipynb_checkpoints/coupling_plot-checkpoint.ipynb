```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
import scienceplots
plt.style.use('science')

df = pd.read_table("coupling_results.txt",delimiter = ' ')
data = np.loadtxt('../ATLAS_Tables/1602.05606_ATLAS_Fig7-b_pp_Tbj_Wb_s_L.txt')
m_T = data[:,0]
obs = data[:,1]
```

```python
import matplotlib.patches as mpatches
fig, axes = plt.subplots(1,2,figsize=(16,4))
ax = axes[0]
ax1 = axes[1]
filt1 = df['res'] == 0
filt2 = df['res'] == 1
sns.kdeplot(x = df.loc[filt1,'mass'],y = df.loc[filt1,'s_L'],fill=False,color='yellow',levels=1000,ax=ax)#,fill = True,color = 'yellow',bw_adjust=0.5,ax = ax) #,levels = 100
ax.set_xlim(df['mass'].min(),df['mass'].max())
ax.set_ylim(df['s_L'].min(),df['s_L'].max())
ax.set_xlabel(r'$m_T$ [GeV]')
ax.set_ylabel(r'$sin(\theta_L)$')
excluded_patch = mpatches.Rectangle((0, 0), 1, 1, facecolor='yellow', edgecolor='yellow', label='LHC 95\% CL exclusion')
excluded_patch2 = mpatches.Rectangle((0, 0), 1, 1, facecolor='yellow', edgecolor='yellow', label='LHC 95\% CL exclusion')
ax1.plot(m_T,obs,color = 'red',linewidth = 2,label = '95\% observed limit')
ax1.scatter(df.loc[filt1,'mass'],df.loc[filt1,'s_L'],color = 'yellow',s = 2)
ax1.set_xlim(600,1200)
ax1.set_ylim(0,1)
ax1.set_xlabel(r'$m_T$ [GeV]')
ax1.set_ylabel(r'$sin(\theta_L)$')
ax1.add_patch(excluded_patch)
ax1.legend(loc='lower right')
ax.add_patch(excluded_patch2)
ax.legend(loc='lower right')
fig.savefig('coupling.png')
```

```python
df2 = pd.read_table("coupling_results_2channels.txt",delimiter = ' ')
data2 = np.loadtxt('../ATLAS_Tables/1812.07343_ATLAS_fig8a_pp_Tbq_Wbbq_Singlet_sintheta_L.txt')
m_T2 = data2[:,0]
obs2 = data2[:,1]
filt1 = df2['ratio'] < 1 
filt2 = df2['ratio'] >= 1
fig, axes = plt.subplots(1,2,figsize=(16,4))
ax = axes[0]
ax2 = axes[1]
sns.kdeplot(x = df2.loc[filt2,'mass'],y = df2.loc[filt2,'s_L'],fill = False,levels = 4000, bw_adjust=0.5,color = 'yellow',ax=ax) #,levels = 100 bw_adjust=0.5,
ax.set_xlim(610,1195)
ax.set_ylim(0,1)
ax.set_xlabel(r'$m_T$ [GeV]')
ax.set_ylabel(r'$sin(\theta_L)$')
ax2.set_xlim(602,1195)
ax2.set_ylim(0,1)
ax2.set_xlabel(r'$m_T$ [GeV]')
ax2.set_ylabel(r'$sin(\theta_L)$')
ax2.plot(m_T2,obs2,color='g',label='95\% observed limit')
ax2.plot(m_T,obs,color='r',label='95\% observed limit')
ax2.scatter(df2.loc[filt2,'mass'],df2.loc[filt2,'s_L'],color='yellow',s=4,label='LHC 95\% CL exclusion')
excluded_patch = mpatches.Rectangle((0, 0), 1, 1, facecolor='yellow', edgecolor='yellow', label='LHC 95\% CL exclusion')
ax.add_patch(excluded_patch)
ax.legend(loc='lower right')
ax2.legend(loc='lower right')
```
