```python

fig, axes = plt.subplots(1,2,figsize=(18,7))
ax1 = axes[0]
ax2 = axes[1]

def r_x(m1, m2): return m1/m2
def lambda_func(x, y, z):
    lamda = x ** 4 + y ** 4 + z ** 4 - 2 * x ** 2 * y ** 2 - 2 * x ** 2 * z ** 2 - 2 * y ** 2 * z ** 2
    return lamda
def decay_to_wb(x ,mv_theo): 
    G = 0.64
    PI = 3.14
    MW = 80.377 
    Mb = 4.18
    constant = G**2 / (64 * PI)

    s_l = np.sqrt(x / (constant / (MW ** 2) * np.sqrt(lambda_func(mv_theo, Mb, MW)) * (1 + r_x(MW, mv_theo) ** 2 
          - 2 * r_x(Mb, mv_theo) ** 2 - 2 * r_x(MW, mv_theo) ** 4 + r_x(Mb, mv_theo) ** 4 + r_x(Mb, mv_theo) ** 2 * r_x(MW, mv_theo) ** 2))) 
    return s_l
kappa = []
kappa1 = []
kappa2 = []
kappa3 = []
mass = []
for m in range(1000, 2200, 1):
    kappa.append(decay_to_wb(0.5,m))
    kappa1.append(decay_to_wb(0.3,m))
    kappa2.append(decay_to_wb(0.2,m))
    kappa3.append(decay_to_wb(0.1,m))
    mass.append(m)
#print(kappa)
#mass = np.arange(1000,2200,1)

#df = pd.read_table("coupling_limit_from_class.dat",delimiter = ' ')
df = pd.read_table("coupling_limit_from_class_doublet.dat",delimiter = ' ')


filter1 = (df["width"] < 0.50) & (df["res"] == 0 )

filter2 = (df["res"] == 0) & (df["width"] > 0.50)

filter3 = df["width"] == 0.5 #(df["width"] >= 0.4995) & (df["width"] <= 0.5005)
#filter4 = (df["r"] >= 0.296) & (df["r"] <= 0.303)
filter5 = (df["width"] >= 0.098) & (df["width"] <= 0.102)

filtered_m = df.loc[filter1,"mass"]
filtered_k = df.loc[filter1,"sin_l"]

filtered_m2 = df.loc[filter2,"mass"]
filtered_k2 = df.loc[filter2,"sin_l"]

x = df.loc[filter3,"mass"]
y = df.loc[filter3,"sin_l"]

#f_x = interp1d(x,y,'cubic')

lighter_gray = (0.90, 0.90, 0.90)

sns.kdeplot(x= filtered_m,y=filtered_k,fill = False,levels= 4000,colors = 'yellow',bw_adjust = 0.5, ax=ax1)
sns.kdeplot(x= filtered_m,y=filtered_k,fill = False,levels= 2,colors = 'black',bw_adjust = 0.5,linewidths=0.5, ax=ax1)
sns.kdeplot(x= filtered_m2,y=filtered_k2,fill = False,levels= 4000,colors=[lighter_gray],bw_adjust = 0.5 , ax=ax1)
sns.kdeplot(x= filtered_m2,y=filtered_k2,fill = False,levels= 2,colors = 'black',bw_adjust = 0.5,linewidths=0.5, ax=ax1)

excluded_patch = mpatches.Rectangle((0, 0), 1, 1, facecolor='yellow', edgecolor='black', label='LHC 95\% CL exclusion')
ax1.add_patch(excluded_patch)
excluded_patch = mpatches.Rectangle((0, 0), 1, 1, facecolor='lightgray', edgecolor='black', label=r'$\Gamma_T/m_T > 50\%$')
ax1.add_patch(excluded_patch)

#ax2.plot( df.loc[filter3,"mass"],df.loc[filter3,"sin_l"],marker='o',linestyle='--',markersize=1.2,color='black')

ax2.scatter(df.loc[filter2,"mass"], df.loc[filter2,"sin_l"], color=lighter_gray, s = 3)
ax2.scatter(df.loc[filter1,"mass"], df.loc[filter1,"sin_l"], color='yellow', s = 3)

#ax2.plot(df.loc[filter4,"m"],df.loc[filter4,"k"],marker='o',linestyle=' ',markersize=1.2,color='black')

#ax2.plot(df.loc[filter5,"m"],df.loc[filter5,"k"],marker='o',linestyle=' ',markersize=1.2,color='black')


'''
data1 = np.loadtxt("/home/mohamed/PycharmProjects/TOP/ATLAS_Tables/2305.03401_ATLAS_Fig13a_pp_Tqt_Wb_Zt_Ht_singlet.txt")
m1 = data1[:,0]
obs1 = data1[:,1]
ax2.plot(m1,obs1,color='blue',lw=2,label = 'arXiv:2305.03401 observed limit')


data2 = np.loadtxt("/home/mohamed/PycharmProjects/TOP/ATLAS_Tables/2402.16561_ATLAS_Fig12a_pp_T_Ht_Zt_singlet.txt")
m1 = data2[:,0]
obs1 = data2[:,1]
obs2 = data2[:,2]
ax2.plot(m1,obs1,color='red',lw=2,label = 'arXiv:2402.16561 observed limit')


data3 = np.loadtxt("/home/mohamed/PycharmProjects/TOP/ATLAS_Tables/2201.07045_ATLAS_Fig9_pp_Tbq_Wb_Ht_Zt_singlet.txt")
m1 = data3[:,0]
obs1 = data3[:,1]
obs2 = data3[:,2]
ax2.plot(m1,obs1,color='green',lw=2,label = 'arXiv:2201.07045 observed limit')

data4 = np.loadtxt("/home/mohamed/PycharmProjects/TOP/ATLAS_Tables/2307.07584_ATLAS_Fig9a_pp_T_Wb_Ht_Zt_singlet.txt")
m1 = data4[:,0]
obs1 = data4[:,1]
obs2 = data4[:,2]
ax2.plot(m1,obs1,color='cyan',lw=2,label = 'arXiv:2307.07584 observed limit')
'''

data1 = np.loadtxt("/home/mohamed/PycharmProjects/TOP/ATLAS_Tables/2305.03401_ATLAS_Fig13b_pp_Tqt_Ht_Zt_doublet.txt")
m1 = data1[:,0]
obs1 = data1[:,1]
ax2.plot(m1,obs1,color='blue',lw=2,label = 'arXiv:2305.03401 observed limit')


data2 = np.loadtxt("/home/mohamed/PycharmProjects/TOP/ATLAS_Tables/2307.07584_ATLAS_Fig9b_pp_T_Ht_Zt_doublet.txt")
m1 = data2[:,0]
obs1 = data2[:,1]
obs2 = data2[:,2]
ax2.plot(m1,obs1,color='red',lw=2,label = 'arXiv:2307.07584 observed limit')


ax2.plot(mass,kappa, linestyle = '--',lw = 2,color = 'black')
ax2.plot(mass,kappa1, linestyle = '--',lw = 2,color = 'black')
ax2.plot(mass,kappa2, linestyle = '--',lw = 2,color = 'black')
ax2.plot(mass,kappa3, linestyle = '--',lw = 2,color = 'black')
#ax2.scatter(df.loc[filter3,"m"],df.loc[filter3,"k"],color='green')

ax1.set_xlim(1000,2130)
ax1.set_ylim(0,1)
ax1.set_xlabel(r"$m_T$ [GeV]")
ax1.set_ylabel(r"$k_T$",fontsize = 16)
ax2.text(1500, 0.79, r'$\Gamma_T/m_T = 50\%$', fontsize=9, color='black',fontweight='bold',rotation=-26)
ax2.text(1600, 0.57, r'$\Gamma_T/m_T = 30\%$', fontsize=9, color='black',fontweight='bold',rotation=-19)
ax2.text(1700, 0.44, r'$\Gamma_T/m_T = 20\%$', fontsize=9, color='black',fontweight='bold',rotation=-15)
ax2.text(1800, 0.30, r'$\Gamma_T/m_T = 10\%$', fontsize=9, color='black',fontweight='bold',rotation=-9)
ax2.set_xlim(1000,2130)
ax2.set_ylim(0,1)
ax2.set_xlabel(r"$m_T$ [GeV]")
ax2.set_ylabel(r"$k_T$",fontsize = 16)
ax1.legend(loc='lower right')
ax2.legend(loc='lower right')
fig.suptitle("Doublet vector-like T", fontsize=22, fontweight='bold')
#fig.suptitle("Singlet vector-like T", fontsize=22, fontweight='bold')
plt.savefig('kappa_doublet_final.png')
```
